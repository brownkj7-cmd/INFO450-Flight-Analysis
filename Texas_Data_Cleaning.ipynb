```python

# Count missing distance values
missing_distance = df[df["Distance"].isnull()]

print("Missing distances:", len(missing_distance))

# Check whether each origin-destination route
# has a consistent recorded distance
route_distances = df.groupby(
    ["Origin", "Dest"]
)["Distance"].nunique()

print(
    "Routes with multiple recorded distances:",
    (route_distances > 1).sum()
)
```

```python

# Find the known distance for each route
route_distance_lookup = df.groupby(
    ["Origin", "Dest"]
)["Distance"].transform("first")

# Check how many missing distances can be recovered
recoverable = (
    df["Distance"].isna() &
    route_distance_lookup.notna()
).sum()

unrecoverable = (
    df["Distance"].isna() &
    route_distance_lookup.isna()
).sum()

print("Missing distances:", df["Distance"].isna().sum())
print("Recoverable distances:", recoverable)
print("Unrecoverable distances:", unrecoverable)
```

so after our analysis we are able to see the dataset contains 600 missing distance values approx 1% of the data. Missing valuse are spread out relatively even across the airport,airlines and cancelation status meaning that MCAR is plausible.

We can see that every route with distance recoreded had only one constant distance value and all 600b of the missing values can be recovered by using ditances recovered for the other flights on the same orgin to destianation route.

So during our phase 3 were going to fill the missing values and then use the correlated route distance rather than overall mean by doing this we will preseve the route information and not over or under estimate creating errors.
