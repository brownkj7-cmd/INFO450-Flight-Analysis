## Texas Fligts Reliability and Timing

### Project overview
Our group will study flight delays and cancellations at four Texas airports: Dallas/Fort Worth (DFW), Houston (IAH), Austin (AUS), and San Antonio (SAT). We will use flight records from August 2025 through July 2026 to learn how reliable flights are in this region. Our goal is to find patterns that could help people plan trips and understand why some flights are less reliable than others.

### Research questions

**Question 1 : How reliable is flying out of our region?**  
We will compare the on-time arrival percentage and cancellation rate for DFW, IAH, AUS, and SAT.

**Question 2 : How concentrated are delays?**  
We will check whether a small number of flights, routes, or destinations account for most of the total delay minutes.

**Question 3 : When are delays worst?**  
We will look at delays by month, day of the week, and scheduled departure time, including how delays change across the year.

**Question 6 : Are weekend flights more delayed than weekday flights?**  
We will compare weekend and weekday arrival delays and check whether any difference is large enough to matter in practice.

### Who would use this information?
- **Travelers** could use the results to decide when to fly and which airport to use.
- **Business travel managers** could use the results when planning employee trips and allowing extra travel time.
- **Airlines and airport managers** could use the findings to identify times or routes where delays are more common.

### What we expect to find (before analyzing the data)
These are our predictions.

- **Question 1:** We think DFW and IAH might have lower on-time arrival percentages than AUS and SAT because they are busy airports with many connecting flights. We will test this instead of assuming it is true.
- **Question 2:** We expect a relatively small share of flights or routes to account for a large share of delay minutes, since a few very long delays can add up quickly.
- **Question 3:** We think flights later in the day may have longer delays because earlier delays can affect later flights. We also expect some months to be worse than others.
- **Question 6:** We think weekend and weekday flights may have different average delays because travel patterns change. We are not sure which group will be worse or whether the difference will be meaningful.

### Why this project matters
Flight delays can cause missed connections, extra costs, and changes to people's plans. By comparing airports and looking for patterns in delay times, our group hopes to provide information that helps travelers and businesses make better travel decisions.
